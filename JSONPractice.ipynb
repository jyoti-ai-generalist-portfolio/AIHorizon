1. **Converting Python Dictionary to JSON Object (String)**

```python
#Converting Python Dictionary to JSON Object (String)

myDict = {}
myDict['A'] = "Amitabh Bacchan"
myDict['B'] = "Badshah"
myDict['C'] = "Count of Monte Cristo"
myDict['D'] = "Don"
myDict['E'] = 12

print (myDict)

import json

myJSONString = json.dumps(myDict)
print (myJSONString)

```

2. **Reading a JSON String into Python. If JSON String is an object it gets converted into a dictionary**

```python
#Converting JSON String (Object) to a Python Dictionary

import json

myJSONString = '{"Name":"Jyoti","Age":48,"Gender":"F","IsOkay":true}'
#JSON String is converted to dictionary
myDict = json.loads(myJSONString)
print ("My Name is : " , myDict["Name"])
print ("ISOkay value ", myDict["IsOkay"])
```

**3. Converting JSON String (Array)   -> Python Array**

```python
# Converting JSON String (Array)   -> Python Array

import json

myJSONString = '[1,2,3,4,5]'

myObj = json.loads(myJSONString)
print (myObj)
print (myObj[1])
print (myObj[3])
```

4. **Converting JSON String (array of objects  )  -> Python Array of dictionaries**

```python
# Converting JSON String (array of objects  )  -> Python Array of dictionaries

import json

myJSONString = '[{"Name":"Jyoti","Age":48,"Gender":"F","IsOkay":true},{"Name":"Rishikesh","Age":52,"Gender":"M","IsOkay":false},{"Name":"Jahnavi", "Age":22, "Gender": "F", "Flag" : true}]'

myObj = json.loads(myJSONString)
print (myObj[0])
print (myObj [2])

print (myObj[1].get("Name"))

```

5.  **Writing a python object to file as a JSON**

*   Use dump (object, fileobject) to actually write it
*    Use following Python code to open a file in Write mode and assign the file object to a variable.
  With Open("user_data.json",w) as file:




To see the file in Colab Click on the Folder Icon on left hand side it will show file saved as user_data.json

```python
import json

myUserProfile = { "Name":"Jyoti", "Skills":"Python,Java,C", "Age":48, "Flag": True}

with open("user_data.json","w") as myFile:
  json.dump(myUserProfile, myFile)

print ("File Saved : user_data.json")
```

6. **Reading from a JSON file saved on same directory and converting it to a Python object**

Use :

*   json.load(file)
*   Use python code


with open("user_data.json","r") as file

```python
import json

with open("user_data.json","r") as myFileObject:
  myObj = json.load(myFileObject)

print (myObj)
```

7. **Doing the same with multiple objects inside an array**




```python
import json

#My python object is an array of dictionaries.
# I am converting it to a string of JSON format and writing it to a file
# In Python array means List
myPythonArrayOfDict = [\
    {"Name":"Jyoti","Age":48,"Gender":"F","IsOkay":True},\
     {"Name":"Rishikesh","Age":52,"Gender":"M","IsOkay":False},\
      {"Name":"Jahnavi", "Age":22, "Gender": "F", "Flag" : True}\
                       ]


with open("user_data2.json","w") as myFile1:
  json.dump(myPythonArrayOfDict,myFile1)

print ("Dumped into file user_data2.json")

with open ("user_data2.json","r") as myFile2:
  myNewJSONString = json.load(myFile2)

print (type(myNewJSONString))

print (myNewJSONString[0].get("Name"))
print (myNewJSONString[1].get("Age"))
print (myNewJSONString[2].get("Gender"))

```
