#PROJECT 4 : COMBINING PYTHON PANDAS DATA WORK WITH AI ANALYSIS

PROJECT: Load a public CSV, clean it with Pandas, calculate 5 business KPIs and use Gemini to generate a concise management summary.


This sample dataset contains 100 mock retail transactions across 4 product categories and 4 geographic regions.

 It includes fields like Revenue, COGS (Cost of Goods Sold), Marketing_Spend, Customer_ID, and CSAT_Score to help you calculate and build your Python project.


**5 Recommended Business KPIs for this Dataset**


You can easily calculate the following metrics using pandas:


**Net Profit Margin:**<p>
Formula: (Revenue - COGS - Marketing_Spend) / Revenue * 100 <p>
Purpose: Measures the exact profitability percentage of your sales after basic operational costs.<p>
**Customer Retention Rate (CRR):**<p>
Formula: Group by Customer_ID to count repeat shoppers versus one-time buyers.<p>
Purpose: Determines how effectively the company keeps customers coming back.<p>
**Marketing ROI (Return on Investment):**<p>
Formula: (Revenue - COGS) / Marketing_Spend<p>
Purpose: Analyzes the financial efficiency and revenue performance of marketing expenditures.<p>
**Average Customer Satisfaction (CSAT) Score:**<p>
Formula: df['CSAT_Score'].mean()<p>
Purpose: Monitors customer happiness levels; can be aggregated by region or product category.<p>
**Average Order Value (AOV):**<p>
Formula: df['Revenue'].mean()<p>
Purpose: Tracks the average dollar amount spent per transaction to gauge pricing health.<p>


```python
from google import genai
import json
from google.colab import userdata
from pydantic import BaseModel
import pandas as pd


try:
    input_file_name = "company_performance_data.csv"
    df = pd.read_csv(input_file_name)
    #print ("File loaded successfully")
    #print (df.head())

    # Data cleaning - remove null values and re-write to the same file
    df.fillna(0, inplace=True)
    #print (df)
    df.to_csv(input_file_name,index=False)


    myprompt = """## 1. Persona (Who the AI is)
            > Act as an expert **Data Analyst and BI analyst* with specialized knowledge in **Financial Metrics and Operations**. Your tone should be **Professional**.

            ## 2. Context & Background (Why/What is happening)
            * **The Situation:** The company wants to report Top 5 KPI metrics based on the data provided
            * **Target Audience:** Senior Managers, CFO, CEO
            * **Core Goal:** Report of the Top 5 KPI metrics of the company

            ## 3. Objective & Task (What to execute)
            Please perform the following action(s):
            1. **Task 1: Analyse the given company data**
            2. **Task 2: identify top 5 KPI metrics of this company that can be calculated based on the data**
            3. **Task 3: Prepare a summary report in Text format which will have these Top 5 metrics displayed for senior management review**
            4. **Task 4: Also prepare appropriate graphs for these which can be used as images in a presentation deck**



            ## 4. Constraints & Guardrails (What NOT to do)
            * **Constraint 1:** Do not use data not in the sheet

            ## 5. Output Specification (How it looks)
            Format your response exactly as follows:
            # EXECUTIVE MANAGEMENT SUMMARY

            ## 1. Financial Performance Overview
            * Total Revenue Achieved: $[Insert Calculated Sum of Revenue]
            * Total Gross Profit: $[Insert Revenue - COGS]
            * Average Net Profit Margin: [Insert Average Margin]%

            ## 2. Strategic Insights & Operational Efficiency
            * Best Performing Product Category: [Insert Top Category by Revenue]
            * Most Profitable Regional Market: [Insert Top Region by Net Profit]
            * Marketing Campaign Efficiency (ROI): [Insert Overall ROI Factor]x return on spend.

            ## 3. Customer Success & Loyalty Metrics
            * Customer Satisfaction Index: [Insert Average CSAT]/5.0
            * Repeat Purchase Rate: [Insert % of Unique Customers with >1 Transaction]%

            ## 4. Key Recommendations
            * [Action 1]: Scale up marketing budgets in the [Top Region] region due to higher net margins.
            * [Action 2]: Address lower-performing categories like [Lowest Category] by reviewing product pricing strategies.
              """
    client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))
    print ("Client created successfully")
    uploadedFile = client.files.upload(file=input_file_name)
    print (uploadedFile)
    response = client.models.generate_content( \
              model="gemini-3.5-flash-lite", \
              contents=[uploadedFile,myprompt], \
              )

    print (response.text)



except FileNotFoundError:
  print ("File not found")

```
