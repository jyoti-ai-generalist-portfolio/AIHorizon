To get a structured output from Gemini 3.6 Flash (gemini-3.6-flash), pass a Pydantic model or a JSON Schema directly into the response_schema parameter of the generation_config, and set the response_mime_type to application/json using the official google-genai




```python
from google import genai
from google.colab import userdata
from google.genai import types
from pydantic import BaseModel

# 1 Initialise the genai client
# Automatically picks up GEMINI_API_KEY from environment variables
# In this case its not in ENVIRONMENT VAR since we put it in secret key in collab
client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))
# 2. Define your desired output structure using Pydantic
class Sentiment(BaseModel):
  reviewer_id: str
  sentiment: str
# 3. Upload the csv file to the client for analysis
csv_file = client.files.upload(file="./data.csv")

# 4. Set my prompt
myprompt = "Analyze this CSV data do the sentiment analysis as POSITIVE, NEGATIVE OR NEUTRAL against each record. \
There are 10 rows in this csv. Analyze the sentiment for each and every row without skipping any.\
Provide the output for all rows as JSON"


# 4. Call the model with the schema configuration
response = client.models.generate_content( \
    model="gemini-3.6-flash", \
    contents=[csv_file,myprompt], \
    config=types.GenerateContentConfig( \
        response_mime_type="application/json", \
        response_schema=list[Sentiment],  # Pass the Pydantic class here \
    ))


print (response.text)
```
