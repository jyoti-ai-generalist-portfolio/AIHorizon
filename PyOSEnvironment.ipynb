Method to store API keys using .env File

python-dotenv is a popular Python library used to read key-value pairs from a .env file and set them as environment variables. This approach keeps sensitive configuration data (like API keys, database passwords, and secrets) out of your source code, aligning with standard 12-factor application development practices.



1. Install the packageRun the following command in your terminal

```python
!pip install python-dotenv
```

2. Create a .env file

Place a plain text file named .env in the root directory of your project. Do not include spaces around the = sign:

```python
import os
from dotenv import load_dotenv


# Load variables from the .env file into the system environment
load_dotenv()

# Access the variables using os.getenv()
api_key = os.getenv("API_KEY")
db_url = os.getenv("DATABASE_URL")
port = os.getenv("PORT", "5000")  # '5000' acts as a default fallback value

print(f"API Key: {api_key}")
print(f"Connecting to: {db_url}")
```
