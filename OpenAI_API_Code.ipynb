```python
# This code will not work as Open AI is not free. Although code is similar hence shifted to GROK

import json
from openai import OpenAI
from google.colab import userdata

client = OpenAI(api_key=userdata.get("OPEN_AI_API_KEY"))
myprompt = """ Give me a short summary of how we can use Open AI API in Python
          """


response = client.chat.completions.create(\
            model="gpt-4o",  # You can also use "gpt-4-turbo" or "gpt-3.5-turbo",
            messages=[\
                {"role": "system", "content": "You are a helpful assistant."}, \
                {"role": "user", "content": myprompt} \
            ] \
        )
        # Extract and return the text response
print (response.choices[0].message.content)





```
